## 1. List Characteristics

A Python list is:

- **ordered**, so every element has a position;
- **mutable**, so its contents can be changed;
- able to contain **duplicate values**;
- able to contain values of different types.

Lists are written using square brackets.

````python
list1 = ["apple", "banana", "cherry"]
list2 = [1, 2, 3, 4, 5]
list3 = [True, False, True]
list4 = ["apple", 1, True]
list5 = [1, 2, 3, True, "apples", "bananas", "cherries"]

print(list1)
print(list2)
print("Length of list3:", len(list3))
print(list4)
print(list5)
````

## 2. Indexing

Python begins counting list positions at `0`.

```text
Positive index:  0   1   2   3   4   5   6
Negative index: -7  -6  -5  -4  -3  -2  -1
```

A negative index counts backward from the end.

````python
values = [1, 2, 3, True, "apples", "bananas", "cherries"]

print("First element:", values[0])
print("Fifth element:", values[4])
print("Last element:", values[-1])
print("Third element from the end:", values[-3])
````

Trying to use an index that does not exist causes an `IndexError`.

````python
try:
    print(values[20])
except IndexError:
    print("IndexError: that position does not exist.")
````

## 3. Adding, Removing, and Replacing Items

`append()` changes the existing list and returns `None`. Therefore, do not write `myList = myList.append(value)`.

````python
fruits = ["apples", "bananas", "cherries"]

fruits.append("strawberries")
print("After append:", fruits)

fruits.remove("apples")
print("After remove:", fruits)

fruits[0] = "blueberries"
print("After replacement:", fruits)
````

Other useful methods include `insert()`, `pop()`, `clear()`, `count()`, and `index()`.

````python
numbers = [10, 20, 30]

numbers.insert(1, 15)
removedValue = numbers.pop()

print("Updated list:", numbers)
print("Removed value:", removedValue)
````

## 4. Ranges and List Slicing

A slice uses `list[start:stop:step]`. The stop position is not included.

````python
numbers1 = list(range(1, 100))
numbers2 = list(range(1, 100, 5))

maximumValue = max(numbers1)
minimumValue = min(numbers2)

A = numbers2[0:5]
B = numbers2[0:50:3]
C = numbers2[3:]
D = numbers2[::2]
E = numbers2[::-1]
F = numbers2[-3]
G = numbers2[-3:]

print("Maximum of numbers1:", maximumValue)
print("Minimum of numbers2:", minimumValue)
print("A:", A)
print("B:", B)
print("C:", C)
print("D:", D)
print("E:", E)
print("F:", F)
print("G:", G)
````

## 5. List Exercises

### Exercise 1: Find the Highest Two Distinct Values

### Solution

````python
values = [8, 12, 3, 19, 19, 5, 14]

distinctValues = sorted(set(values))
highestTwo = distinctValues[-2:]

print("Highest two distinct values:", highestTwo)
````

### Exercise 2: Switch the First and Last Elements

### Solution

````python
values = ["a", "b", "c", "d", "e"]

values[0], values[-1] = values[-1], values[0]

print(values)
````

### Exercise 3: Switch the First `n` Entries with the Last `n`

```text
a b c d e f g h i j k l
j k l d e f g h i a b c
```

### Solution

````python
values = list("abcdefghijkl")
n = 3

if 0 <= n <= len(values) // 2:
    switchedValues = values[-n:] + values[n:-n] + values[:n]
else:
    switchedValues = "Error: n must be between 0 and half the list length."

print(switchedValues)
````

### Exercise 4: Reverse a List

### Solution

````python
values = [1, 2, 3, 4, 5]

reversedBySlicing = values[::-1]
reversedByMethod = values.copy()
reversedByMethod.reverse()

print("Using slicing:", reversedBySlicing)
print("Using reverse():", reversedByMethod)
````

## 6. Nested Lists

A nested list contains one or more lists as elements. Nested lists can represent matrices, tables, grids, or grouped records.

````python
nestedList = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print("First row:", nestedList[0])
print("Third value in the first row:", nestedList[0][2])
print("Center value:", nestedList[1][1])
````

### Exercise 5: Locate `21` in a Three-Level Nested List

### Solution

````python
deepList = [
    [[1, 2], [3, 4]],
    [[10, 11], [20, 21, 22]],
    [[30, 31], [40, 41]],
]

location = (1, 1, 1)

print("The location is:", location)
print("The value is:", deepList[1][1][1])
````

## 7. Dictionaries

A dictionary maps a unique key to a value. A value is accessed by its key rather than by a numerical index.

````python
cities = {
    "barcelona": 1,
    "madrid": 2,
    "sevilla": 3,
}

print(cities["barcelona"])
print(cities["madrid"])
print(cities["sevilla"])

cities["bilbao"] = 4
print(cities)
````

Avoid naming a variable `dict`, because `dict` is the name of Python's dictionary constructor.

## 8. Nested Dictionaries

````python
person = {
    "fname": "Jane",
    "lname": "Doe",
    "age": 47,
    "spouse": "John",
    "children": ["Alice", "Bob", "Charlie"],
    "pets": {
        "dog": "Buddy",
        "cat": "Whiskers",
    },
}

print(person["fname"])
print(person["age"])
print(person["children"][2])
print(person["pets"]["cat"])
````

## 9. Sorting Dictionaries

````python
myDictionary = {
    "apple": 2,
    "banana": 5,
    "cherry": 3,
    "date": 4,
}

sortedByKeys = dict(sorted(myDictionary.items()))
sortedByValues = dict(
    sorted(myDictionary.items(), key=lambda item: item[1])
)

print("Sorted by keys:", sortedByKeys)
print("Sorted by values:", sortedByValues)
````

## 12. Gradebook Exercise

Store the exam table in a nested dictionary. `None` represents a missing grade.

The program should retrieve a requested exam grade, report the class size, calculate one student's average, and determine whether the student passes using an average of at least 50.

````python
gradebook = {
    "Charlie": {"Exam1": 94, "Exam2": 96, "Exam3": 0, "Exam4": 91},
    "Sam": {"Exam1": 100, "Exam2": 98, "Exam3": 87, "Exam4": 97},
    "Blair": {"Exam1": 89, "Exam2": 0, "Exam3": 89, "Exam4": 77},
    "Karen": {"Exam1": 82, "Exam2": 92, "Exam3": 81, "Exam4": None},
    "Sarah": {"Exam1": 70, "Exam2": 30, "Exam3": 80, "Exam4": 92},
}

requestedStudent = "Sarah"
requestedExam = "Exam3"

requestedGrade = gradebook[requestedStudent][requestedExam]
studentGrades = [
    grade
    for grade in gradebook[requestedStudent].values()
    if grade is not None
]
studentAverage = sum(studentGrades) / len(studentGrades)
passed = studentAverage >= 50

print(f"{requestedStudent}'s {requestedExam} grade:", requestedGrade)
print("Number of students:", len(gradebook))
print(f"{requestedStudent}'s average: {studentAverage:.2f}")
print("Passed:", passed)
````

### Reusable Gradebook Query

````python
def studentReport(records, studentName, examName):
    if studentName not in records:
        return "Student not found."

    if examName not in records[studentName]:
        return "Exam not found."

    grades = [
        grade
        for grade in records[studentName].values()
        if grade is not None
    ]
    average = sum(grades) / len(grades)

    return {
        "requestedGrade": records[studentName][examName],
        "classSize": len(records),
        "average": round(average, 2),
        "passed": average >= 50,
    }

print(studentReport(gradebook, "Karen", "Exam4"))
````
