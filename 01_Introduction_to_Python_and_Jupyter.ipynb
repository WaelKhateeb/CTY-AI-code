## 5. Jupyter Notebooks and Python Scripts

| Feature | Jupyter Notebook | Python Script |
|---|---|---|
| File extension | `.ipynb` | `.py` |
| Execution | One cell at a time | Usually top to bottom |
| Text and explanations | Markdown cells | Mostly comments and documentation |
| Output | Appears below each code cell | Appears in a terminal or output window |
| Best for | Learning, experiments, data analysis | Complete programs and reusable modules |

A notebook is especially useful while learning because explanations, code, and results remain together.

## 6. Markdown Cells and Code Cells

## 8. Visual Studio Code Workflow

A simple class workflow is:

1. Create a folder for the course.
2. Open that folder in Visual Studio Code.
3. Create a file ending in `.ipynb`.
4. Select a Python kernel.
5. Add Markdown and code cells.
6. Run cells using the run button or `Shift + Enter`.
7. Save the notebook regularly.
8. Keep file names clear and descriptive.

A useful folder structure is:

```text
YourName/
    notebooks/
    python_files/
    exercises/
```
