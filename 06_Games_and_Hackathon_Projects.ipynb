# Project 1: Tic-Tac-Toe

The board uses positions `0` through `8`.

```text
0 | 1 | 2
---------
3 | 4 | 5
---------
6 | 7 | 8
```

````python
def createTicTacToeBoard():
    return [" "] * 9


def displayTicTacToeBoard(board):
    for rowStart in range(0, 9, 3):
        print(" | ".join(board[rowStart:rowStart + 3]))

        if rowStart < 6:
            print("-" * 9)


def ticTacToeWinner(board):
    winningLines = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6),
    ]

    for first, second, third in winningLines:
        if board[first] != " " and board[first] == board[second] == board[third]:
            return board[first]

    if " " not in board:
        return "Draw"

    return None


def placeTicTacToeMove(board, position, player):
    if position not in range(9):
        return False

    if board[position] != " ":
        return False

    board[position] = player
    return True
````

### Sample Completed Game

````python
board = createTicTacToeBoard()
moves = [
    ("X", 0),
    ("O", 1),
    ("X", 4),
    ("O", 2),
    ("X", 8),
]

for player, position in moves:
    placeTicTacToeMove(board, position, player)

displayTicTacToeBoard(board)
print("Result:", ticTacToeWinner(board))
````

### Interactive Tic-Tac-Toe Function

The function is defined but not called automatically, so the notebook remains fully executable.

````python
def playTicTacToe():
    board = createTicTacToeBoard()
    currentPlayer = "X"

    while ticTacToeWinner(board) is None:
        displayTicTacToeBoard(board)

        try:
            position = int(
                input(f"Player {currentPlayer}, choose a position from 0 to 8: ")
            )
        except ValueError:
            print("Enter an integer from 0 to 8.")
            continue

        if not placeTicTacToeMove(board, position, currentPlayer):
            print("That move is not available.")
            continue

        currentPlayer = "O" if currentPlayer == "X" else "X"

    displayTicTacToeBoard(board)
    print("Result:", ticTacToeWinner(board))
````

# Project 2: Blackjack

This simplified version gives the player two cards and the dealer two cards. Aces count as 11 unless that would make the hand exceed 21.

````python
def blackjackCardValue(card):
    if card in {"J", "Q", "K"}:
        return 10

    if card == "A":
        return 11

    return int(card)


def blackjackHandValue(hand):
    total = sum(blackjackCardValue(card) for card in hand)
    aceCount = hand.count("A")

    while total > 21 and aceCount > 0:
        total -= 10
        aceCount -= 1

    return total


def blackjackResult(playerHand, dealerHand):
    playerTotal = blackjackHandValue(playerHand)
    dealerTotal = blackjackHandValue(dealerHand)

    if playerTotal > 21:
        return "Dealer wins"
    if dealerTotal > 21:
        return "Player wins"
    if playerTotal > dealerTotal:
        return "Player wins"
    if dealerTotal > playerTotal:
        return "Dealer wins"

    return "Tie"


playerHand = ["A", "K"]
dealerHand = ["10", "9"]

print("Player:", playerHand, blackjackHandValue(playerHand))
print("Dealer:", dealerHand, blackjackHandValue(dealerHand))
print(blackjackResult(playerHand, dealerHand))
````

# Project 3: Connect Four

The board contains six rows and seven columns. A player wins by connecting four matching pieces horizontally, vertically, or diagonally.

````python
def createConnectFourBoard(rows=6, columns=7):
    return [["." for column in range(columns)] for row in range(rows)]


def displayConnectFourBoard(board):
    for row in board:
        print(" ".join(row))


def dropConnectFourPiece(board, column, piece):
    if column < 0 or column >= len(board[0]):
        return False

    for row in range(len(board) - 1, -1, -1):
        if board[row][column] == ".":
            board[row][column] = piece
            return True

    return False


def connectFourWinner(board, piece):
    rows = len(board)
    columns = len(board[0])

    for row in range(rows):
        for column in range(columns):
            directions = [(0, 1), (1, 0), (1, 1), (1, -1)]

            for rowStep, columnStep in directions:
                positions = [
                    (
                        row + step * rowStep,
                        column + step * columnStep,
                    )
                    for step in range(4)
                ]

                if all(
                    0 <= currentRow < rows
                    and 0 <= currentColumn < columns
                    and board[currentRow][currentColumn] == piece
                    for currentRow, currentColumn in positions
                ):
                    return True

    return False
````

````python
connectBoard = createConnectFourBoard()

for column in [0, 1, 2, 3]:
    dropConnectFourPiece(connectBoard, column, "X")

displayConnectFourBoard(connectBoard)
print("X wins:", connectFourWinner(connectBoard, "X"))
````

# Project 4: Roulette

This classroom version supports a straight-number bet and an even-or-odd bet.

````python
def roulettePayout(winningNumber, betType, betValue, wager):
    if winningNumber < 0 or winningNumber > 36:
        return "Error: winning number must be from 0 to 36."

    if betType == "number":
        if int(betValue) == winningNumber:
            return wager * 36
        return 0

    if betType == "parity":
        if winningNumber == 0:
            return 0

        winningParity = "even" if winningNumber % 2 == 0 else "odd"

        if str(betValue).lower() == winningParity:
            return wager * 2
        return 0

    return "Error: unsupported bet type."


print("Number bet payout:", roulettePayout(17, "number", 17, 5))
print("Parity bet payout:", roulettePayout(24, "parity", "even", 10))
````

# Project 5: Hangman

````python
def hangmanDisplay(word, guessedLetters):
    return " ".join(
        letter if letter in guessedLetters else "_"
        for letter in word
    )


def hangmanWrongGuesses(word, guessedLetters):
    return sorted(
        letter
        for letter in guessedLetters
        if letter not in word
    )


def hangmanWon(word, guessedLetters):
    return all(letter in guessedLetters for letter in word)


secretWord = "python"
guessedLetters = set()

for guess in ["p", "a", "y", "t", "h", "o", "n"]:
    guessedLetters.add(guess)
    print(hangmanDisplay(secretWord, guessedLetters))

print("Wrong guesses:", hangmanWrongGuesses(secretWord, guessedLetters))
print("Won:", hangmanWon(secretWord, guessedLetters))
````

# Project 6: Bingo

A bingo card has five rows and five columns. This example checks complete rows, columns, and the two main diagonals.

````python
import random

def createBingoCard(seed=7):
    generator = random.Random(seed)
    ranges = [
        range(1, 16),
        range(16, 31),
        range(31, 46),
        range(46, 61),
        range(61, 76),
    ]

    columns = [
        generator.sample(list(numberRange), 5)
        for numberRange in ranges
    ]

    card = [
        [columns[column][row] for column in range(5)]
        for row in range(5)
    ]
    card[2][2] = "FREE"
    return card


def markBingoCard(card, calledNumber):
    return [
        [
            "X" if value == calledNumber or value == "FREE" else value
            for value in row
        ]
        for row in card
    ]


def hasBingo(card):
    rowWin = any(all(value == "X" for value in row) for row in card)

    columnWin = any(
        all(card[row][column] == "X" for row in range(5))
        for column in range(5)
    )

    diagonalWin = all(card[index][index] == "X" for index in range(5))
    reverseDiagonalWin = all(
        card[index][4 - index] == "X"
        for index in range(5)
    )

    return rowWin or columnWin or diagonalWin or reverseDiagonalWin
````

````python
bingoCard = createBingoCard()
firstRowNumbers = bingoCard[0].copy()

for number in firstRowNumbers:
    bingoCard = markBingoCard(bingoCard, number)

for row in bingoCard:
    print(row)

print("Bingo:", hasBingo(bingoCard))
````

# Project 7: Slot Machine

The random-number generator uses a fixed seed so the notebook produces the same demonstration each time.

````python
def slotMachineSpin(seed=None):
    generator = random.Random(seed)
    symbols = ["Cherry", "Lemon", "Bell", "Seven"]
    return [generator.choice(symbols) for reel in range(3)]


def slotMachinePayout(spin, wager):
    if spin[0] == spin[1] == spin[2]:
        return wager * 10

    if len(set(spin)) == 2:
        return wager * 2

    return 0


spin = slotMachineSpin(seed=11)
wager = 5

print("Spin:", spin)
print("Payout:", slotMachinePayout(spin, wager))
````
