### Exercise 1: Age Checks

Take a classmate's age and return `True` or `False` for each statement:

1. The age is between 10 and 13.
2. The age is less than half of Albert Einstein's age of 76.
3. The age squared is greater than 200.
4. `(age + 50) / 2` is greater than 13.

### Exercise 2: Polygon Angles

For a polygon with `n` sides:

- sum of interior angles: `(n - 2) × 180`;
- each angle in a regular polygon: `sum / n`.

### Solution

```python
numberOfSides = 6

sumOfInteriorAngles = (numberOfSides - 2) * 180
eachInteriorAngle = sumOfInteriorAngles / numberOfSides

print("Sum of interior angles:", sumOfInteriorAngles)
print("Each angle in a regular polygon:", eachInteriorAngle)
```

## 4. Error Handling with `try` and `except`

Place code that may fail inside `try`. When a matching error occurs, Python moves to `except` instead of stopping the entire program.

```python
def convertToNumber(text):
    try:
        return float(text)
    except ValueError:
        return "Error: enter a valid number."

print(convertToNumber("12.5"))
print(convertToNumber("twelve"))
```

## 5. Number Classification Exercise

Read a number and report whether it is:

- positive, negative, or zero;
- even or odd when it is an integer;
- an integer or decimal;
- divisible by 3 when it is an integer;
- less than 1000 in absolute value.

### Solution

```python
def classifyNumber(userValue):
    try:
        number = float(userValue)
    except ValueError:
        return ["Error: enter a valid number."]

    results = []

    if number > 0:
        results.append("The number is positive.")
    elif number < 0:
        results.append("The number is negative.")
    else:
        results.append("The number is zero.")

    if number.is_integer():
        integerNumber = int(number)
        results.append("The number is an integer.")

        if integerNumber % 2 == 0:
            results.append("The number is even.")
        else:
            results.append("The number is odd.")

        if integerNumber % 3 == 0:
            results.append("The number is divisible by 3.")
        else:
            results.append("The number is not divisible by 3.")
    else:
        results.append("The number is decimal.")
        results.append("Even, odd, and divisibility by 3 are checked only for integers.")

    if abs(number) < 1000:
        results.append("Its absolute value is less than 1000.")
    else:
        results.append("Its absolute value is at least 1000.")

    return results

for line in classifyNumber("-27"):
    print(line)
```

## 6. Grade Classifier

Use this scale:

- 90–100: A
- 80–89: B
- 70–79: C
- 60–69: D
- 0–59: F

Values outside 0–100 must produce an error.

### Solution

```python
def classifyGrade(userValue):
    try:
        grade = float(userValue)
    except ValueError:
        return "Error: the grade must be a number."

    if grade < 0 or grade > 100:
        return "Error: the grade must be between 0 and 100."
    elif grade >= 90:
        return "A"
    elif grade >= 80:
        return "B"
    elif grade >= 70:
        return "C"
    elif grade >= 60:
        return "D"
    else:
        return "F"

for testGrade in ["95", "84", "72", "63", "40", "105", "hello"]:
    print(testGrade, "->", classifyGrade(testGrade))
```

## 7. Temperature Conversion Menu

Menu:

1. Kelvin to Fahrenheit  
2. Fahrenheit to Kelvin  
3. Fahrenheit to Celsius  
4. Celsius to Fahrenheit  

An invalid choice must produce an error message.

### Solution

```python
def convertTemperature(choice, temperature):
    try:
        choice = int(choice)
        temperature = float(temperature)
    except ValueError:
        return "Error: choice and temperature must be numeric."

    if choice == 1:
        result = (temperature - 273.15) * 9 / 5 + 32
        return f"{temperature:.2f} K = {result:.2f} °F"
    elif choice == 2:
        result = (temperature - 32) * 5 / 9 + 273.15
        return f"{temperature:.2f} °F = {result:.2f} K"
    elif choice == 3:
        result = (temperature - 32) * 5 / 9
        return f"{temperature:.2f} °F = {result:.2f} °C"
    elif choice == 4:
        result = temperature * 9 / 5 + 32
        return f"{temperature:.2f} °C = {result:.2f} °F"
    else:
        return "Error: choose an option from 1 to 4."

print(convertTemperature(1, 300))
print(convertTemperature(2, 68))
print(convertTemperature(3, 86))
print(convertTemperature(4, 25))
print(convertTemperature(7, 25))
```

## 8. ATM Exercise

The ATM:

- accepts the valid PIN `1234`;
- allows at most three PIN attempts;
- displays `PIN locked` after three incorrect attempts;
- dispenses the smallest possible number of `$10`, `$5`, and `$1` notes.

The solution below avoids requiring keyboard input by accepting three PIN attempts as function arguments.

```python
def atmWithdrawal(firstPin, secondPin, thirdPin, amount):
    validPin = "1234"

    if str(firstPin) == validPin:
        pinAccepted = True
    elif str(secondPin) == validPin:
        pinAccepted = True
    elif str(thirdPin) == validPin:
        pinAccepted = True
    else:
        pinAccepted = False

    if not pinAccepted:
        return "PIN locked"

    try:
        amount = int(amount)
    except ValueError:
        return "Error: the withdrawal amount must be an integer."

    if amount <= 0:
        return "Error: the withdrawal amount must be positive."

    tenDollarNotes = amount // 10
    remainder = amount % 10
    fiveDollarNotes = remainder // 5
    oneDollarNotes = remainder % 5

    return (
        f"{tenDollarNotes} note(s) of $10, "
        f"{fiveDollarNotes} note(s) of $5, and "
        f"{oneDollarNotes} note(s) of $1"
    )

print(atmWithdrawal("9999", "1234", "0000", 36))
print(atmWithdrawal("1111", "2222", "3333", 36))
```
