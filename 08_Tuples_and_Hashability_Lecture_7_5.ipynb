## 1. Creating Tuples

Lists use square brackets. Tuples normally use parentheses, although commas are what actually create the tuple.

```python
tupleA = (4, 3, 2)
myTuple = (1, 2, 3, "a", "b")
anotherTuple = 1, 2, 3
singleItemTuple = (5,)

print(tupleA)
print(myTuple)
print(anotherTuple)
print(singleItemTuple)
print(type(anotherTuple))
print(type(singleItemTuple))
```

Without the comma, `(5)` is simply the integer `5`, not a tuple.

```python
notATuple = (5)
actualTuple = (5,)

print(type(notATuple))
print(type(actualTuple))
```

## 2. Lists Are Mutable; Tuples Are Immutable

A list can be changed after creation. A tuple cannot have an element replaced, appended, or removed.

```python
valuesList = [10, 20, 30]
valuesList[1] = 99
valuesList.append(40)

print(valuesList)
```

```python
valuesTuple = (10, 20, 30)

try:
    valuesTuple[1] = 99
except TypeError as error:
    print(type(error).__name__ + ":", error)
```

A tuple is immutable, not mutable. It may still contain a mutable object such as a list, but the tuple's own element positions cannot be replaced.

```python
mixedTuple = ([1, 2], "fixed")
mixedTuple[0].append(3)

print(mixedTuple)
```

## 3. Tuple Methods

Tuples provide two commonly used methods:

- `count(value)` reports how many times a value appears;
- `index(value)` reports the first matching position.

```python
values = (4, 7, 4, 9, 4, 11)

print("Number of 4s:", values.count(4))
print("First position of 9:", values.index(9))
```

## 4. Tuple Unpacking

Unpacking assigns tuple elements to separate variables.

```python
studentRecord = ("Maya", 14, "Programming")
studentName, studentAge, courseName = studentRecord

print(studentName)
print(studentAge)
print(courseName)
```

## 5. Returning Multiple Values from a Function

Python packages multiple returned values into a tuple.

```python
def rectangleMeasurements(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    return area, perimeter


result = rectangleMeasurements(8, 5)
area, perimeter = result

print("Returned object:", result)
print("Returned type:", type(result))
print("Area:", area)
print("Perimeter:", perimeter)
```

## 6. Choose a List or a Tuple

| Situation | Better choice |
|---|---|
| Values must be added, removed, or replaced | List |
| Fixed ordered collection | Tuple |
| Sequence must be a dictionary key | Tuple, when all elements are hashable |
| Sequence must be a set element | Tuple, when all elements are hashable |

```python
choices = {
    "Modify elements later": "list",
    "Fixed RGB color": "tuple",
    "Coordinate used as a dictionary key": "tuple",
}

for situation, answer in choices.items():
    print(situation, "->", answer)
```

## 7. Memory and Timing

Tuples are often slightly smaller and can be slightly faster to construct than equivalent lists. The difference is usually minor, so choose based primarily on whether the collection should change.

```python
import sys
import timeit

listExample = [1, 2, 3, 4, 5]
tupleExample = (1, 2, 3, 4, 5)

print("List size:", sys.getsizeof(listExample), "bytes")
print("Tuple size:", sys.getsizeof(tupleExample), "bytes")

listTime = timeit.timeit("[1, 2, 3, 4, 5]", number=100_000)
tupleTime = timeit.timeit("(1, 2, 3, 4, 5)", number=100_000)

print(f"List construction time: {listTime:.6f} seconds")
print(f"Tuple construction time: {tupleTime:.6f} seconds")
```

Timing results depend on the computer and Python version. They should be treated as a small local experiment rather than a universal constant.

## 8. Hashability

A hash value is an integer used by hash-based structures such as dictionaries and sets.

A hashable object must:

- have a hash value that remains constant during its lifetime;
- support equality comparison;
- satisfy the rule that equal objects have equal hashes.

Immutable built-in values such as integers and strings are hashable. A tuple is hashable only when every element inside it is hashable.

```python
print("Hash of 42:", hash(42))
print("Hash of a string:", hash("Real Python"))

numbers = (1, 2, 3)
print("Hash of a tuple:", hash(numbers))
```

String hashes may differ between separate Python sessions because Python randomizes them for security.

```python
try:
    print(hash([1, 2, 3]))
except TypeError as error:
    print(type(error).__name__ + ":", error)
```

```python
hashableTuple = (1, "a", 3.5)
tupleContainingList = (1, [2, 3])

print("Hashable tuple:", hash(hashableTuple))

try:
    print(hash(tupleContainingList))
except TypeError as error:
    print(type(error).__name__ + ":", error)
```

## 9. Tuples as Dictionary Keys

Coordinates are a common example.

```python
locations = {
    (0, 0): "Origin",
    (3, 4): "Point A",
    (-2, 5): "Point B",
}

print(locations[(3, 4)])
```

## 10. Tuples in Sets

```python
visitedCoordinates = {
    (0, 0),
    (1, 2),
    (3, 4),
}

visitedCoordinates.add((5, 6))

print(visitedCoordinates)
```

## 11. Hashability Exercise

Test several values and report whether each is hashable.

### Solution

```python
objects = [
    42,
    "Python",
    (1, 2, 3),
    [1, 2, 3],
    {"a": 1},
    (1, [2, 3]),
]

for value in objects:
    try:
        hash(value)
        result = "hashable"
    except TypeError:
        result = "not hashable"

    print(repr(value), "->", result)
```
