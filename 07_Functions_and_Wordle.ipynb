## 4. Exercise: Sum Every Value in a List

### Solution

```python
def listTotal(values):
    total = 0

    for value in values:
        total += value

    return total


print(listTotal([3, 7, 10, 5]))
```

## 5. Exercise: Multiply Every Value in a List

### Solution

```python
def listProduct(values):
    product = 1

    for value in values:
        product *= value

    return product


print(listProduct([2, 3, 4, 5]))
```

The product of an empty list is defined here as `1`, matching the identity value for multiplication.

## 7. Exercise: Maximum and Average of a Float List

The solutions below do not use `max()` or `sum()`.

```python
def listMaximum(values):
    if len(values) == 0:
        raise ValueError("The list cannot be empty.")

    currentMaximum = values[0]

    for value in values[1:]:
        if value > currentMaximum:
            currentMaximum = value

    return currentMaximum


def listAverage(values):
    if len(values) == 0:
        raise ValueError("The list cannot be empty.")

    total = 0

    for value in values:
        total += value

    return total / len(values)


measurements = [2.5, 7.1, 3.8, 9.4, 6.0]

print("Maximum:", listMaximum(measurements))
print("Average:", listAverage(measurements))
```

## 8. Exercise: Pangram Checker

A pangram contains every letter of the English alphabet at least once.

### Solution

```python
import string


def isPangram(sentence):
    lettersInSentence = {
        character.lower()
        for character in sentence
        if character.isalpha()
    }

    return set(string.ascii_lowercase).issubset(lettersInSentence)


example = "The quick brown fox jumps over the lazy dog"

print(isPangram(example))
print(isPangram("Python is readable"))
```

## 9. Exercise: Greatest Common Divisor and Least Common Multiple

The Euclidean algorithm repeatedly replaces `(a, b)` with `(b, a % b)`.

### Solution

```python
def greatestCommonDivisor(firstNumber, secondNumber):
    firstNumber = abs(firstNumber)
    secondNumber = abs(secondNumber)

    while secondNumber != 0:
        firstNumber, secondNumber = (
            secondNumber,
            firstNumber % secondNumber,
        )

    return firstNumber


def leastCommonMultiple(firstNumber, secondNumber):
    if firstNumber == 0 or secondNumber == 0:
        return 0

    divisor = greatestCommonDivisor(firstNumber, secondNumber)
    return abs(firstNumber * secondNumber) // divisor


print("GCD of 48 and 18:", greatestCommonDivisor(48, 18))
print("LCM of 48 and 18:", leastCommonMultiple(48, 18))
```

## 10. Exercise: Draw a Hollow Right Triangle

The input controls the height. The final row is completely filled.

### Solution

```python
def drawTriangle(height):
    if height < 2:
        return "*"

    lines = []

    for row in range(1, height + 1):
        if row == 1:
            lines.append("*")
        elif row == height:
            lines.append("*" * row)
        else:
            insideSpaces = " " * (row - 2)
            lines.append("*" + insideSpaces + "*")

    return "\n".join(lines)


print(drawTriangle(6))
```

## 11. Wordle Project

For each letter:

- `G` means the letter is correct and in the correct position;
- `Y` means the letter appears elsewhere in the answer;
- `-` means the letter does not receive a match.

Duplicate letters must be handled carefully. A letter in the answer cannot be matched more times than it appears.

```python
from collections import Counter


def scoreWordleGuess(answer, guess):
    answer = answer.lower()
    guess = guess.lower()

    if len(answer) != len(guess):
        raise ValueError("The answer and guess must have the same length.")

    result = ["-"] * len(answer)
    remainingLetters = Counter()

    for index, guessedLetter in enumerate(guess):
        if guessedLetter == answer[index]:
            result[index] = "G"
        else:
            remainingLetters[answer[index]] += 1

    for index, guessedLetter in enumerate(guess):
        if result[index] == "G":
            continue

        if remainingLetters[guessedLetter] > 0:
            result[index] = "Y"
            remainingLetters[guessedLetter] -= 1

    return result


def displayWordleResult(guess, result):
    return " ".join(
        f"[{status}:{letter.upper()}]"
        for letter, status in zip(guess, result)
    )
```

```python
answer = "apple"

for guess in ["paper", "ample", "apple"]:
    result = scoreWordleGuess(answer, guess)
    print(displayWordleResult(guess, result))
```

### Complete Wordle Round Function

The function accepts a prepared list of guesses, making it easy to test without keyboard input.

```python
def playWordleRound(answer, guesses, maximumAttempts=6):
    history = []

    for attemptNumber, guess in enumerate(
        guesses[:maximumAttempts],
        start=1,
    ):
        result = scoreWordleGuess(answer, guess)
        history.append((attemptNumber, guess, result))

        if guess.lower() == answer.lower():
            return {
                "won": True,
                "attempts": attemptNumber,
                "history": history,
            }

    return {
        "won": False,
        "attempts": len(history),
        "history": history,
    }


game = playWordleRound(
    "crane",
    ["stare", "grape", "crane"],
)

print("Won:", game["won"])
print("Attempts:", game["attempts"])

for attemptNumber, guess, result in game["history"]:
    print(attemptNumber, displayWordleResult(guess, result))
```
