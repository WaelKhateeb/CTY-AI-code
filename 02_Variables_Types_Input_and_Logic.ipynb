### Exercise 1: Boston Marathon Unit Conversion

Use a marathon distance of `26.2` miles. Convert it to:

- kilometers;
- feet;
- inches;
- centimeters;
- millimeters.

### Solution

```python
marathonMiles = 26.2

marathonKilometers = marathonMiles * 1.60934
marathonFeet = marathonMiles * 5280
marathonInches = marathonFeet * 12
marathonCentimeters = marathonInches * 2.54
marathonMillimeters = marathonCentimeters * 10

print(f"Kilometers:  {marathonKilometers:.3f}")
print(f"Feet:        {marathonFeet:,.1f}")
print(f"Inches:      {marathonInches:,.1f}")
print(f"Centimeters: {marathonCentimeters:,.1f}")
print(f"Millimeters: {marathonMillimeters:,.1f}")
```

The string `"5"` is text. The integer `5` is a number.

### Exercise 3: Passport Card

Collect a person's name, age, school, and city, then display an organized passport-style card.

Interactive programs would use `input()`. The function below accepts values directly so it can be tested repeatedly.

### Solution

```python
def printPassport(name, age, school, city):
    print("=" * 36)
    print("STUDENT PASSPORT".center(36))
    print("=" * 36)
    print("Name", name, sep=": ")
    print("Age", age, sep=": ")
    print("School", school, sep=": ")
    print("City", city, sep=": ")
    print("=" * 36)

printPassport("Alice Johnson", 13, "CTY Academy", "Baltimore")
```

### Exercise 9: Circle Measurements

For radius `r`:

- circumference is `2 × π × r`;
- area is `π × r²`.

### Solution

```python
import math

def circleMeasurements(radius):
    circumference = 2 * math.pi * radius
    area = math.pi * radius ** 2
    return circumference, area

circumference, area = circleMeasurements(5)
print(f"Circumference: {circumference:.3f}")
print(f"Area: {area:.3f}")
```

## 9. Boolean Operators

Python uses:

- `and`: both conditions must be true;
- `or`: at least one condition must be true;
- `not`: reverses a Boolean value;
- XOR: exactly one input is true.

NAND and NOR are formed by applying `not` to AND and OR.

```python
def AND(a, b):
    return a and b

def OR(a, b):
    return a or b

def NOT(a):
    return not a

def XOR(a, b):
    return a != b

def NAND(a, b):
    return not (a and b)

def NOR(a, b):
    return not (a or b)

def XNOR(a, b):
    return a == b
```

### Truth Table: `(A and B) or C`

### Solution

```python
print("A B C | (A and B) or C")
print("-" * 25)

for A in [False, True]:
    for B in [False, True]:
        for C in [False, True]:
            result = (A and B) or C
            print(int(A), int(B), int(C), "|", int(result))
```

### Basic Gate Truth Tables

### Solution

```python
print("A B | AND OR XOR NAND NOR XNOR")
print("-" * 34)

for A in [False, True]:
    for B in [False, True]:
        values = [
            AND(A, B), OR(A, B), XOR(A, B),
            NAND(A, B), NOR(A, B), XNOR(A, B)
        ]
        print(int(A), int(B), "|", *(int(value) for value in values))
```
